Q11: Identify and remove outliers in the km_driven column using the IQR method

```python
import pandas as pd

data = pd.read_csv('https://drive.google.com/uc?id=1iKy23bMtEQShF_weneRNnYrFmzvpPOI3')

Q1 = data['km_driven'].quantile(0.25)
Q3 = data['km_driven'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

data_no_outliers = data[(data['km_driven'] >= lower_bound) & (data['km_driven'] <= upper_bound)]

print("Original data shape:", data.shape)
print("Data shape after removing outliers:", data_no_outliers.shape)
```

Q14: Which bike names are priced significantly above the average price for their manufacturing year

```python
import pandas as pd

data = pd.read_csv('https://drive.google.com/uc?id=1iKy23bMtEQShF_weneRNnYrFmzvpPOI3')

avg_price_by_year = data.groupby('year')['selling_price'].mean()

std_price_by_year = data.groupby('year')['selling_price'].std()

def is_overpriced(row):
    year = row['year']
    price = row['selling_price']
    avg_price = avg_price_by_year.get(year, 0)
    std_price = std_price_by_year.get(year, 0)

    return price > avg_price + std_price

data['overpriced'] = data.apply(is_overpriced, axis=1)

overpriced_bikes = data[data['overpriced']]['name'].unique()

print("Bikes priced significantly above average for their manufacturing year:")
for bike in overpriced_bikes:
    print(bike)
```

Q7: What are the most popular product reviews

```python
import pandas as pd
from collections import Counter

data = pd.read_csv('/content/amazon.csv')

all_product_names = ' '.join(data['product_name'].astype(str).tolist())

words = all_product_names.lower().split()

word_counts = Counter(words)

stop_words = ['for', 'with', 'and', 'of', 'the', 'in', 'to', 'a', 'is', 'it', 'that', 'on', 'you', 'this', 'be', 'as', 'are', 'or', 'not', 'your', 'from', 'have', 'an', 'by', 'at', 'i']
for word in stop_words:
    if word in word_counts:
        del word_counts[word]

top_n = 10
most_common_keywords = word_counts.most_common(top_n)

print(f"Top {top_n} Most Popular Product Keywords:")
for keyword, frequency in most_common_keywords:
    print(f"{keyword}: {frequency}")
```

Q8:  What is the correlation between discounted_price and rating

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv('/content/amazon.csv')

data['discounted_price'] = pd.to_numeric(data['discounted_price'].str.replace(',', '').str.replace('₹', ''), errors='coerce')
data['rating'] = pd.to_numeric(data['rating'], errors='coerce')


correlation = data['discounted_price'].corr(data['rating'])

print(f"Correlation between discounted_price and rating: {correlation:.2f}")

plt.figure(figsize=(10, 6))
sns.scatterplot(x='discounted_price', y='rating', data=data)
plt.title('Relationship between Discounted Price and Rating')
plt.xlabel('Discounted Price')
plt.ylabel('Rating')
plt.show()
```

Q4:  Which artist has the highest number of tracks in the dataset? Display the count of tracks for each artist using
a countplot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.read_csv('/content/spotify.csv')

artist_counts = data['Artist'].value_counts()
artist_with_most_tracks = artist_counts.index[0]
most_tracks_count = artist_counts.iloc[0]

print(f"The artist with the highest number of tracks is: {artist_with_most_tracks}")
print(f"Number of tracks: {most_tracks_count}")

plt.figure(figsize=(12, 6))
sns.countplot(y='Artist', data=data, order=artist_counts.index)
plt.title('Number of Tracks by Artist')
plt.xlabel('Count')
plt.ylabel('Artist')
plt.tight_layout()
plt.show()
```

Q6: Among the top 5 most popular artists, which artist has the highest popularity on average? Calculate and
display the average popularity for each artist

```python
import pandas as pd

data = pd.read_csv('/content/spotify.csv')

top_5_artists = data.groupby('Artist')['Popularity'].sum().sort_values(ascending=False).head(5).index

average_popularity = data[data['Artist'].isin(top_5_artists)].groupby('Artist')['Popularity'].mean().sort_values(ascending=False)

print("Average Popularity for Top 5 Most Popular Artists:")
print(average_popularity)

artist_with_highest_avg_popularity = average_popularity.index[0]
highest_avg_popularity = average_popularity.iloc[0]

print(f"\nThe artist with the highest average popularity is: {artist_with_highest_avg_popularity}")
print(f"Average Popularity: {highest_avg_popularity}")
```

Q9: Does the duration of tracks vary significantly across different artists? Explore this visually using a box plot or
violin plot

```python
!pip install seaborn
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('/content/spotify.csv')

plt.figure(figsize=(12, 6))
sns.boxplot(x='Artist', y='Duration (ms)', data=data)
plt.title('Distribution of Track Duration Across Artists (Box Plot)')
plt.xlabel('Artist')
plt.ylabel('Duration (ms)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.violinplot(x='Artist', y='Duration (ms)', data=data)
plt.title('Distribution of Track Duration Across Artists (Violin Plot)')
plt.xlabel('Artist')
plt.ylabel('Duration (ms)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
