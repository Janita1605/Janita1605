Q3: Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution

```python
def bernoulli_pdf(x, p):
    """
    Calculate the probability distribution function (PDF) of a Bernoulli distribution.

    Args:
    x (int): The outcome (0 or 1).
    p (float): The probability of success (0 <= p <= 1).

    Returns:
    float: The probability of the outcome x.
    """
    if x not in [0, 1]:
        raise ValueError("x must be 0 or 1")
    if not (0 <= p <= 1):
        raise ValueError("p must be between 0 and 1 inclusive")

    if x == 1:
        return p  # Probability of success (x = 1)
    else:
        return 1 - p  # Probability of failure (x = 0)

# Example usage:
p = 0.7  # Probability of success
x = 1    # Outcome (either 0 or 1)

pdf_value = bernoulli_pdf(x, p)
print(f"The probability of x = {x} is: {pdf_value}")
```

Q10:  Implement the Central Limit Theorem (CLT) using Python for a non-normal distribution.

```python
import numpy as np
import matplotlib.pyplot as plt

def central_limit_theorem(population_size, sample_size, num_samples):
    """
    Apply the Central Limit Theorem to a non-normal population distribution (Exponential distribution).

    Args:
    population_size (int): The size of the population to sample from.
    sample_size (int): The size of each sample.
    num_samples (int): The number of samples to draw.

    Returns:
    np.ndarray: An array of sample means.
    """
    # Generate a non-normal distribution (Exponential distribution)
    population = np.random.exponential(scale=2.0, size=population_size)

    # List to store the sample means
    sample_means = []

    # Draw multiple samples and calculate their means
    for _ in range(num_samples):
        sample = np.random.choice(population, size=sample_size, replace=False)
        sample_means.append(np.mean(sample))

    return np.array(sample_means)

# Parameters
population_size = 10000  # Size of the population (Exponential distribution)
sample_size = 30        # Size of each sample
num_samples = 1000      # Number of samples to draw

# Apply CLT
sample_means = central_limit_theorem(population_size, sample_size, num_samples)

# Plot the histogram of the sample means
plt.hist(sample_means, bins=30, edgecolor='black', density=True)
plt.title("Central Limit Theorem (CLT) - Sample Means Distribution")
plt.xlabel("Sample Mean")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

Q17: Generate random variables and calculate their corresponding probabilities using the binomial distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

def binomial_distribution(n, p, size):
    """
    Generate random variables and calculate their corresponding probabilities
    using the binomial distribution.

    Args:
    n (int): Number of trials
    p (float): Probability of success on each trial
    size (int): Number of random variables to generate

    Returns:
    np.ndarray: Random variables
    np.ndarray: Corresponding probabilities
    """
    # Generate random variables from a binomial distribution
    random_variables = np.random.binomial(n, p, size)

    # Calculate the corresponding probabilities (PMF) for each possible outcome
    k_values = np.arange(0, n+1)  # Possible outcomes (0, 1, ..., n)
    probabilities = binom.pmf(k_values, n, p)  # Binomial PMF

    return random_variables, k_values, probabilities

# Parameters for binomial distribution
n = 10  # Number of trials
p = 0.5  # Probability of success on each trial
size = 1000  # Number of random variables to generate

# Generate random variables and calculate their corresponding probabilities
random_variables, k_values, probabilities = binomial_distribution(n, p, size)

# Plot histogram of the random variables
plt.hist(random_variables, bins=np.arange(-0.5, n+1.5, 1), density=True, alpha=0.7, label='Random Variables')
plt.plot(k_values, probabilities, 'bo-', label='Binomial PMF', lw=2)
plt.title(f"Binomial Distribution (n={n}, p={p})")
plt.xlabel("Number of Successes (k)")
plt.ylabel("Probability")
plt.legend()
plt.grid(True)
plt.show()
```

Q18:  Write a Python program to calculate the Z-score for a given data point and compare it to a standard normal
distribution

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def calculate_z_score(data_point, mean, std_dev):
    """
    Calculate the Z-score for a given data point.

    Args:
    data_point (float): The data point for which to calculate the Z-score.
    mean (float): The mean of the distribution.
    std_dev (float): The standard deviation of the distribution.

    Returns:
    float: The calculated Z-score.
    """
    z_score = (data_point - mean) / std_dev
    return z_score

def compare_to_standard_normal(z_score):
    """
    Compare the calculated Z-score to the standard normal distribution (mean=0, std=1).

    Args:
    z_score (float): The calculated Z-score to compare.

    Returns:
    float: The probability corresponding to the Z-score from the standard normal distribution.
    """
    # Calculate the cumulative probability for the Z-score from the standard normal distribution
    prob = norm.cdf(z_score)
    return prob

# Parameters
mean = 50  # Example mean of the data
std_dev = 10  # Example standard deviation of the data
data_point = 65  # Example data point to calculate the Z-score for

# Calculate Z-score
z_score = calculate_z_score(data_point, mean, std_dev)

# Compare the Z-score to the standard normal distribution
probability = compare_to_standard_normal(z_score)

# Output the results
print(f"Data point: {data_point}")
print(f"Z-score: {z_score:.2f}")
print(f"Probability (CDF) for Z-score: {probability:.4f}")

# Plot the standard normal distribution and highlight the Z-score
x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, 0, 1)

plt.plot(x, y, label="Standard Normal Distribution (μ=0, σ=1)", color='blue')
plt.fill_between(x, y, where=(x <= z_score), color='orange', alpha=0.5, label=f"Z-score = {z_score:.2f}")

plt.axvline(x=z_score, color='red', linestyle='--', label=f'Z-score = {z_score:.2f}')
plt.title("Standard Normal Distribution and Z-score")
plt.xlabel("Z")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```

Q19:  Implement hypothesis testing using Z-statistics for a sample dataset

```python
import numpy as np
import scipy.stats as stats

def z_test(sample_data, population_mean, population_std, significance_level):
    """
    Perform a Z-test for the sample data against a population mean.

    Args:
    sample_data (array-like): Sample data to perform the test on.
    population_mean (float): The population mean (null hypothesis).
    population_std (float): The population standard deviation.
    significance_level (float): The significance level for the test.

    Returns:
    tuple: Z-statistic, p-value, decision (reject/fail to reject H0).
    """
    # Calculate sample statistics
    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)

    # Calculate the Z-statistic
    z_statistic = (sample_mean - population_mean) / (population_std / np.sqrt(sample_size))

    # Calculate the p-value for a two-tailed test
    p_value = 2 * (1 - stats.norm.cdf(abs(z_statistic)))

    # Compare p-value with the significance level
    if p_value < significance_level:
        decision = "Reject the null hypothesis"
    else:
        decision = "Fail to reject the null hypothesis"

    return z_statistic, p_value, decision

# Sample data (e.g., heights of 30 people in cm)
sample_data = [172, 168, 175, 169, 178, 174, 171, 169, 177, 173,
               175, 176, 174, 171, 170, 169, 165, 179, 182, 180,
               174, 173, 169, 168, 176, 172, 174, 175, 170, 171]

# Known population parameters
population_mean = 170  # Null hypothesis mean (cm)
population_std = 10    # Population standard deviation (cm)

# Significance level
alpha = 0.05

# Perform Z-test
z_statistic, p_value, decision = z_test(sample_data, population_mean, population_std, alpha)

# Output results
print(f"Z-statistic: {z_statistic:.2f}")
print(f"P-value: {p_value:.4f}")
print(f"Decision: {decision}")

```

Q20:  Create a confidence interval for a dataset using Python and interpret the result

```python
import numpy as np
import scipy.stats as stats

def confidence_interval(sample_data, confidence_level=0.95):
    """
    Calculate the confidence interval for a given sample data.

    Args:
    sample_data (array-like): The sample data for which the confidence interval is calculated.
    confidence_level (float): The confidence level (default is 0.95).

    Returns:
    tuple: Lower and upper bounds of the confidence interval.
    """
    # Calculate sample statistics
    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)
    sample_std = np.std(sample_data, ddof=1)  # Sample standard deviation (Bessel's correction)

    # Calculate the Z-value or t-value based on sample size
    alpha = 1 - confidence_level
    critical_value = stats.t.ppf(1 - alpha / 2, df=sample_size - 1)  # t-distribution critical value

    # Calculate the margin of error
    margin_of_error = critical_value * (sample_std / np.sqrt(sample_size))

    # Calculate the confidence interval
    ci_lower = sample_mean - margin_of_error
    ci_upper = sample_mean + margin_of_error

    return ci_lower, ci_upper

# Example sample data (e.g., heights in cm of a sample of people)
sample_data = [172, 168, 175, 169, 178, 174, 171, 169, 177, 173,
               175, 176, 174, 171, 170, 169, 165, 179, 182, 180,
               174, 173, 169, 168, 176, 172, 174, 175, 170, 171]

# Calculate the confidence interval for the sample data
confidence_level = 0.95
ci_lower, ci_upper = confidence_interval(sample_data, confidence_level)

# Output the results
print(f"Sample Mean: {np.mean(sample_data):.2f} cm")
print(f"Confidence Interval ({confidence_level*100}%): ({ci_lower:.2f}, {ci_upper:.2f}) cm")
```

Q22: Write a Python script to calculate and visualize the probability density function (PDF) of a normal distribution

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

def plot_normal_pdf(mu, sigma, x_range):
    """
    Plots the Probability Density Function (PDF) of a normal distribution.

    Args:
    mu (float): Mean of the normal distribution.
    sigma (float): Standard deviation of the normal distribution.
    x_range (tuple): Range of x-values to plot the PDF.
    """
    # Generate x-values
    x = np.linspace(x_range[0], x_range[1], 1000)

    # Calculate the PDF using the normal distribution formula
    pdf_values = stats.norm.pdf(x, loc=mu, scale=sigma)

    # Plot the PDF
    plt.figure(figsize=(8, 5))
    plt.plot(x, pdf_values, label=f'Normal Distribution\n$\mu={mu}$, $\sigma={sigma}$', color='blue')
    plt.title(f'Probability Density Function of Normal Distribution\n$\mu={mu}$, $\sigma={sigma}$')
    plt.xlabel('x')
    plt.ylabel('Probability Density')
    plt.grid(True)
    plt.legend()
    plt.show()

# Parameters for the normal distribution
mu = 0        # Mean
sigma = 1     # Standard deviation

# Define the range for x-values (from -4 to +4 standard deviations from the mean)
x_range = (-4*sigma + mu, 4*sigma + mu)

# Plot the PDF
plot_normal_pdf(mu, sigma, x_range)
```

Q23:  Use Python to calculate and interpret the cumulative distribution function (CDF) of a Poisson distribution

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

def plot_poisson_cdf(lmbda, x_range):
    """
    Plots the Cumulative Distribution Function (CDF) of a Poisson distribution.

    Args:
    lmbda (float): The mean number of occurrences (lambda) of the Poisson distribution.
    x_range (tuple): Range of x-values to plot the CDF.
    """
    # Generate x-values (discrete values)
    x = np.arange(0, x_range+1)

    # Calculate the CDF using the Poisson distribution
    cdf_values = stats.poisson.cdf(x, mu=lmbda)

    # Plot the CDF
    plt.figure(figsize=(8, 5))
    plt.step(x, cdf_values, where='post', label=f'Poisson Distribution\n$\lambda={lmbda}$', color='blue', linestyle='-', marker='o')
    plt.title(f'Cumulative Distribution Function (CDF) of Poisson Distribution\n$\lambda={lmbda}$')
    plt.xlabel('x')
    plt.ylabel('Cumulative Probability')
    plt.grid(True)
    plt.legend()
    plt.show()

# Parameters for the Poisson distribution
lmbda = 5    # The mean number of occurrences (lambda)

# Define the range for x-values (from 0 to 15 for illustration)
x_range = 15

# Plot the CDF of the Poisson distribution
plot_poisson_cdf(lmbda, x_range)
```

Q27:  Implement Z-score normalization on a dataset and visualize its transformation

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate a dataset from a normal distribution
np.random.seed(42)  # For reproducibility
data = np.random.normal(loc=50, scale=15, size=1000)  # Mean = 50, Std = 15

# Calculate the mean and standard deviation of the dataset
mean_data = np.mean(data)
std_data = np.std(data)

# Apply Z-score normalization
z_score_normalized_data = (data - mean_data) / std_data

# Visualize the original data and normalized data
plt.figure(figsize=(12, 6))

# Plot original data histogram
plt.subplot(1, 2, 1)
plt.hist(data, bins=30, color='lightblue', edgecolor='black')
plt.title('Original Data')
plt.xlabel('Value')
plt.ylabel('Frequency')

# Plot normalized data histogram
plt.subplot(1, 2, 2)
plt.hist(z_score_normalized_data, bins=30, color='salmon', edgecolor='black')
plt.title('Z-score Normalized Data')
plt.xlabel('Z-score')
plt.ylabel('Frequency')

# Display the plots
plt.tight_layout()
plt.show()

# Print the mean and standard deviation before and after normalization
print(f"Original Mean: {mean_data:.2f}, Original Std Dev: {std_data:.2f}")
print(f"Normalized Mean: {np.mean(z_score_normalized_data):.2f}, Normalized Std Dev: {np.std(z_score_normalized_data):.2f}")
```

Q28: Write a Python function to calculate the skewness and kurtosis of a dataset generated from a normal
distribution

```python
import numpy as np
import scipy.stats as stats

def calculate_skewness_kurtosis(data):
    """
    Calculate the skewness and kurtosis of a dataset.

    Args:
    data (np.ndarray): The input dataset to calculate skewness and kurtosis.

    Returns:
    tuple: (skewness, kurtosis)
    """
    # Calculate skewness
    skew = stats.skew(data)

    # Calculate kurtosis (excess kurtosis is used by default)
    kurt = stats.kurtosis(data)

    return skew, kurt

# Generate a dataset from a normal distribution
np.random.seed(42)  # For reproducibility
data = np.random.normal(loc=0, scale=1, size=1000)  # Mean = 0, Std = 1

# Calculate skewness and kurtosis
skewness, kurtosis = calculate_skewness_kurtosis(data)

# Print the results
print(f"Skewness of the dataset: {skewness:.4f}")
print(f"Kurtosis of the dataset: {kurtosis:.4f}")
```
